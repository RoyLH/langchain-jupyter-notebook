```typescript
import { Ollama } from "@langchain/community/llms/ollama";

const ollama = new Ollama({
  baseUrl: "http://localhost:11434",
  model: "llama3.2",
});

async function chatWithOllama(input: string) {
  try {
    // 功能：生成文本输出，传入一个字符串，Ollama 返回结果。
    const response = await ollama.call(input);

    console.log("用户输入：", input);
    console.log("Ollama 回复：", response);
  } catch (error) {
    console.error("请求失败：", error);
  }
}


async function main() {
  console.log("与 Ollama 开始对话...\n");

  const questions = [
    "What is LangChain?",
    "How does it integrate with Deno?",
    "What are LLMs?",
  ];

  for (const question of questions) {
    await chatWithOllama(question);
    console.log("\n");
  }
}

main().catch((err) => console.error("主函数错误:", err));
```

```typescript
import { Ollama } from "@langchain/community/llms/ollama";

const ollama = new Ollama({
  baseUrl: "http://localhost:11434",
  model: "llama3.2",
});

// 功能：用于批量请求生成结果，并返回更详细的输出结构。
const responses = await ollama.generate([
  "Tell me a joke.",
  "What is LangChain?"
]);

console.log("Ollama 回复：", responses);

responses.generations.forEach((generation, index) => {
  console.log(`请求 ${index + 1}:`, generation[0].text);
});
```

```typescript
import { Ollama } from "@langchain/community/llms/ollama";

const ollama = new Ollama({
  baseUrl: "http://localhost:11434",
  model: "llama3.2",
  streaming: true, // 启用流式输出
});

console.log("Ollama 流式输出：");


// 调用模型并处理流式输出
async function streamOllamaOutput() {
  console.log("开始流式输出...\n");

  let fullText = ""; // 用于拼接完整输出内容

  const response = await ollama.call("Tell me a story about AI.", {
    callbacks: [
      {
        handleLLMNewToken(token: string) {
          process.stdout.write(token); // 实时输出每个 token
          fullText += token; // 拼接完整文本
        },
      },
    ],
  });

  console.log("\n\n完整输出内容：", response);
  console.log("\n\n完整输出内容：", fullText);
}

streamOllamaOutput().catch((err) => console.error("发生错误:", err));
```

```typescript
import { Ollama } from "@langchain/community/llms/ollama";
import { HumanMessage } from "@langchain/core/messages";
import { StringOutputParser } from "@langchain/core/output_parsers";

const ollamaModel = new Ollama({
  baseUrl: "http://localhost:11434",
  model: "llama3.2",
});

// 通过 StringOutputParser 等工具解析输出，再通过自定义函数进一步处理。
const outputParser = new StringOutputParser();

// pipe 方法：将 Ollama 输出与后续处理器串联起来，实现链式调用。
const simpleChain = ollamaModel.pipe(outputParser);

(async () => {

  // 使用 HumanMessage 封装用户输入
  const messages = [
    new HumanMessage("Tell me a joke"),
  ];
  
  // const result = await simpleChain.invoke([
  //   new HumanMessage("Tell me a joke"),
  // ]);

  // 调用链式任务
  const result = await simpleChain.invoke(
    messages[0].text
  ); // 取出 HumanMessage.text 传入

  
  console.log("Ollama 回复：", result);
})();
```
